## Step 17 — Build IQR Outlier Detection Function
**Leveraged and Reused from: Code Reuse Session 1 — Prompt 17**

```python
# define a function to detect outliers in any numeric variable using the IQR method
def find_outliers_IQR(data):
    # calculate Q1 (25th percentile)
    q1 = data.quantile(0.25)
    # calculate Q3 (75th percentile)
    q3 = data.quantile(0.75)
    # calculate the inter-quartile range
    IQR = q3 - q1
    # return values outside the lower fence (Q1-1.5*IQR) or upper fence (Q3+1.5*IQR)
    outliers = data[((data < (q1 - 1.5 * IQR)) | (data > (q3 + 1.5 * IQR)))]
    return outliers
```
